## 3️⃣ BUILD LIFECYCLE + CALIBRATION

```python
# ============================
# 3️⃣ BUILD LIFECYCLE + CALIBRATION
# ============================

print("🔨 Calibrating k and alpha...")

# Actual results
actual_results = get_actual_all_vintages_amount(df_raw)

# DISB_TOTAL map
loan_disb = df_raw.groupby(["PRODUCT_TYPE", "RISK_SCORE", CFG["orig_date"], CFG["loan"]])[CFG["disb"]].first()
disb_total_by_vintage = loan_disb.groupby(level=[0, 1, 2]).sum().to_dict()

# Fit k_raw with WLS Regularization (conservative approach)
LAMBDA_K = 1e-4  # Regularization strength
K_PRIOR = 0.0    # Prior value (bias toward 0 for conservative forecast)

k_raw_by_mob, weight_by_mob, _ = fit_k_raw(
    actual_results=actual_results,
    matrices_by_mob=matrices_by_mob,
    parent_fallback=parent_fallback,
    states=BUCKETS_CANON,
    s30_states=BUCKETS_30P,
    include_co=True,
    denom_mode="disb",
    disb_total_by_vintage=disb_total_by_vintage,
    weight_mode="equal",       # Equal weight for all vintages
    method="wls_reg",          # Regularized WLS for stability
    lambda_k=LAMBDA_K,         # Regularization parameter
    k_prior=K_PRIOR,           # Prior value
    min_obs=5,
    fallback_k=1.0,
    fallback_weight=0.0,
    return_detail=True,
)

print(f"   K values: {len(k_raw_by_mob)} MOBs")

# Smooth k
mob_min = min(k_raw_by_mob.keys()) if k_raw_by_mob else 0
mob_max = max(k_raw_by_mob.keys()) if k_raw_by_mob else 0
k_smooth_by_mob, _, _ = smooth_k(k_raw_by_mob, weight_by_mob, mob_min, mob_max)

# Fit alpha
alpha, k_final_by_mob, _ = fit_alpha(
    actual_results=actual_results,
    matrices_by_mob=matrices_by_mob,
    parent_fallback=parent_fallback,
    states=BUCKETS_CANON,
    s30_states=BUCKETS_30P,
    k_smooth_by_mob=k_smooth_by_mob,
    mob_target=min(MAX_MOB, mob_max) if mob_max else MAX_MOB,
    include_co=True,
)

print(f"   Alpha: {alpha:.4f}")
print(f"   K_final: {len(k_final_by_mob)} MOBs")
```

```python
# Forecast với k_final
forecast_calibrated = forecast_all_vintages_partial_step(
    actual_results=actual_results,
    matrices_by_mob=matrices_by_mob,
    parent_fallback=parent_fallback,
    max_mob=MAX_MOB,
    k_by_mob=k_final_by_mob,
    states=BUCKETS_CANON,
)

# Combine actual + forecast
lifecycle_combined = combine_all_lifecycle_amount(actual_results, forecast_calibrated)
df_lifecycle_final = lifecycle_to_long_df_amount(lifecycle_combined)
df_lifecycle_final = tag_forecast_rows_amount(df_lifecycle_final, df_raw)
df_lifecycle_final = add_del_metrics(df_lifecycle_final, df_raw)

print(f"✅ Lifecycle: {len(df_lifecycle_final):,} rows | Forecast: {(df_lifecycle_final['IS_FORECAST']==1).sum():,}")
```

### 7.1 Tại sao cần hệ số K?

**Vấn đề với Markov Chain thuần túy:**
- Transition Matrix chỉ dự đoán xác suất chuyển đổi trạng thái
- Không tính đến các yếu tố kinh tế vĩ mô, seasonality
- Có thể over/under-estimate DEL rates

**Hệ số K điều chỉnh:**
- K > 1: Model đang under-estimate → cần scale up
- K < 1: Model đang over-estimate → cần scale down
- K ≈ 1: Model dự đoán tốt

### 7.2 Model Accuracy Metrics (MAE, MAPE, RMSE)

```python
# ============================================================
# 7.2 CALCULATE MODEL ACCURACY METRICS
# ============================================================

def calculate_metrics(actual, forecast):
    """Calculate MAE, MAPE, RMSE"""
    actual = np.array(actual)
    forecast = np.array(forecast)
    
    # Remove zeros for MAPE calculation
    mask = actual != 0
    
    mae = np.mean(np.abs(actual - forecast))
    rmse = np.sqrt(np.mean((actual - forecast) ** 2))
    
    if mask.sum() > 0:
        mape = np.mean(np.abs((actual[mask] - forecast[mask]) / actual[mask])) * 100
    else:
        mape = np.nan
    
    return mae, mape, rmse

# Prepare data for comparison
# Filter only rows where we have both actual and forecast
df_eval = df_lifecycle_final[df_lifecycle_final['IS_FORECAST'] == 0].copy()

# We need to compare forecast vs actual for the same cohort/MOB
# This requires running forecast WITHOUT using actual data (backtest)

print("📊 Preparing backtest data...")

# For demonstration, we'll compare DEL30+ rates
if 'DEL30_RATE' in df_eval.columns:
    actual_del30 = df_eval.groupby('MOB')['DEL30_RATE'].mean()
    print(f"\n📈 Average DEL30+ Rate by MOB (Actual):")
    print(actual_del30.head(10))
else:
    print("⚠️ DEL30_RATE column not found")
```

```python
# ============================================================
# 7.3 BACKTEST: Compare Forecast vs Actual
# ============================================================

print("🔄 Running backtest comparison...")

# Get cohorts that have actual data at multiple MOBs
backtest_results = []

for (product, score, vintage), actual_data in actual_results.items():
    max_actual_mob = max(actual_data.keys())
    
    if max_actual_mob < 6:  # Need at least 6 MOBs for meaningful comparison
        continue
    
    # For each MOB, compare actual vs what forecast would have predicted
    for mob in range(3, max_actual_mob + 1):
        if mob not in actual_data:
            continue
        
        actual_amounts = actual_data[mob]
        
        # actual_amounts is a pd.Series, not dict
        # Calculate DEL30+ from actual
        del30_actual = 0
        for s in BUCKETS_30P:
            if s in actual_amounts.index:
                del30_actual += actual_amounts[s]
        
        total_actual = actual_amounts.sum()
        
        if total_actual > 0:
            del30_rate_actual = del30_actual / total_actual
        else:
            del30_rate_actual = 0
        
        backtest_results.append({
            'product': product,
            'score': score,
            'vintage': vintage,
            'mob': mob,
            'del30_actual': del30_actual,
            'total_actual': total_actual,
            'del30_rate': del30_rate_actual,
        })

df_backtest = pd.DataFrame(backtest_results)
print(f"✅ Backtest data: {len(df_backtest):,} observations")
print(f"   Products: {df_backtest['product'].nunique()}")
print(f"   MOB range: {df_backtest['mob'].min()} - {df_backtest['mob'].max()}")
```

```python
# ============================================================
# COHORT COMPARISON CONFIG
# ============================================================

TARGET_COHORT = '2023-12'  # Cohort cần so sánh
START_MOB = 3  # Forecast bắt đầu từ MOB này

print(f'🔍 Tìm cohort: {TARGET_COHORT}')
print(f'   Forecast từ MOB: {START_MOB}')
```

```python
# ============================================================
# TÌM COHORT VÀ CHỌN SEGMENT
# ============================================================

# Tìm tất cả keys có vintage = TARGET_COHORT
matching_keys = []
for key in actual_results.keys():
    product, score, vintage = key
    vintage_dt = pd.to_datetime(vintage)
    target_dt = pd.to_datetime(TARGET_COHORT)
    if vintage_dt.year == target_dt.year and vintage_dt.month == target_dt.month:
        max_mob = max(actual_results[key].keys())
        matching_keys.append((key, max_mob))
        print(f'   Found: {product}/{score}/{vintage} - max MOB: {max_mob}')

if not matching_keys:
    print(f'❌ Không tìm thấy cohort {TARGET_COHORT}!')
    print('\n📊 Các vintage có sẵn (10 gần nhất):')
    vintages = set()
    for key in actual_results.keys():
        vintages.add(pd.to_datetime(key[2]).strftime('%Y-%m'))
    for v in sorted(vintages)[-10:]:
        print(f'   {v}')
else:
    print(f'\n✅ Tìm thấy {len(matching_keys)} segments cho cohort {TARGET_COHORT}')
```

```python
# ============================================================
# CHỌN SEGMENT PHÙ HỢP - AGGREGATE THEO PRODUCT
# ============================================================

# Aggregate tất cả scores trong cùng 1 product cho vintage này
TARGET_PRODUCT = None  # None = tất cả products, hoặc chỉ định 'C', 'S', 'T', etc.

if matching_keys:
    # Group by product
    products_in_cohort = {}
    for key, max_mob in matching_keys:
        product, score, vintage = key
        if product not in products_in_cohort:
            products_in_cohort[product] = []
        products_in_cohort[product].append((key, max_mob))
    
    print(f'\n📊 Products trong cohort {TARGET_COHORT}:')
    for prod, keys in products_in_cohort.items():
        print(f'   {prod}: {len(keys)} segments')
    
    # Chọn product (nếu TARGET_PRODUCT = None, lấy tất cả)
    if TARGET_PRODUCT:
        selected_products = [TARGET_PRODUCT] if TARGET_PRODUCT in products_in_cohort else []
    else:
        selected_products = list(products_in_cohort.keys())
    
    print(f'\n✅ Selected products: {selected_products}')
    
    # Aggregate actual_results theo product
    aggregated_actual = {}  # {mob: pd.Series of EAD by state}
    aggregated_disb = 0
    
    for prod in selected_products:
        for key, max_mob in products_in_cohort[prod]:
            # Add DISB_TOTAL
            aggregated_disb += disb_total_by_vintage.get(key, 0)
            
            # Add actual EAD by MOB
            mob_dict = actual_results[key]
            for mob, ead_series in mob_dict.items():
                if mob not in aggregated_actual:
                    aggregated_actual[mob] = ead_series.copy()
                else:
                    # Sum EAD by state
                    aggregated_actual[mob] = aggregated_actual[mob].add(ead_series, fill_value=0)
    
    print(f'\n📊 Aggregated DISB_TOTAL: {aggregated_disb:,.0f}')
    print(f'   MOBs available: {sorted(aggregated_actual.keys())}')
    
    # Gán lại cho các biến cũ để code phía dưới hoạt động
    disb_total = aggregated_disb
    mob_dict = aggregated_actual
    selected_key = True  # Flag để code phía dưới chạy
    product = 'ALL' if not TARGET_PRODUCT else TARGET_PRODUCT
    score = 'ALL'
else:
    print('❌ Không có segment phù hợp')
    selected_key = None
```

```python
# ============================================================
# TÍNH DEL30+ CHO ACTUAL, FORECAST NO K, FORECAST WITH K
# ============================================================

if selected_key:
    # mob_dict và disb_total đã được aggregate ở cell trước
    print(f'📊 DISB_TOTAL: {disb_total:,.0f}')
    
    if disb_total <= 0:
        print('❌ DISB_TOTAL = 0, không thể tính DEL30_PCT')
    else:
        # Kiểm tra START_MOB có trong data không
        available_mobs = sorted(mob_dict.keys())
        if START_MOB not in mob_dict:
            print(f'⚠️ MOB {START_MOB} không có trong actual data')
            print(f'   Available MOBs: {available_mobs}')
            START_MOB = min(available_mobs)
            print(f'   → Sử dụng START_MOB = {START_MOB}')
        
        initial_ead = mob_dict[START_MOB]
        print(f'\n📊 Initial EAD at MOB {START_MOB} (Aggregated):')
        print(f'   Total EAD: {initial_ead.sum():,.0f}')
        del30_initial = sum(initial_ead[b] if b in initial_ead.index else 0 for b in BUCKETS_30P)
        print(f'   DEL30_AMT: {del30_initial:,.0f}')
        print(f'   DEL30_PCT: {del30_initial/disb_total*100:.2f}%')
```

```python
# ============================================================
# FORECAST VỚI K VÀ KHÔNG K - AGGREGATE TẤT CẢ SEGMENTS
# ============================================================

if selected_key and disb_total > 0:
    # Aggregate forecast từ tất cả segments
    fc_with_k = {}  # {mob: pd.Series}
    fc_no_k = {}    # {mob: pd.Series}
    
    k_no_k = {m: 1.0 for m in range(1, MAX_MOB + 1)}
    
    for prod in selected_products:
        for key, max_mob in products_in_cohort[prod]:
            seg_product, seg_score, seg_vintage = key
            seg_mob_dict = actual_results[key]
            
            # Lấy initial_ead cho segment này
            if START_MOB in seg_mob_dict:
                seg_initial_ead = seg_mob_dict[START_MOB]
            else:
                # Fallback to earliest MOB
                earliest_mob = min(seg_mob_dict.keys())
                seg_initial_ead = seg_mob_dict[earliest_mob]
            
            # Forecast với K
            seg_fc_with_k = forecast_segment_partial_step(
                matrices_by_mob=matrices_by_mob,
                parent_fallback=parent_fallback,
                product=seg_product,
                score=seg_score,
                start_mob=START_MOB,
                initial_ead=seg_initial_ead,
                max_mob=MAX_MOB,
                k_by_mob=k_final_by_mob,
                states=BUCKETS_CANON,
            )
            
            # Forecast không K
            seg_fc_no_k = forecast_segment_partial_step(
                matrices_by_mob=matrices_by_mob,
                parent_fallback=parent_fallback,
                product=seg_product,
                score=seg_score,
                start_mob=START_MOB,
                initial_ead=seg_initial_ead,
                max_mob=MAX_MOB,
                k_by_mob=k_no_k,
                states=BUCKETS_CANON,
            )
            
            # Aggregate
            for mob, ead_series in seg_fc_with_k.items():
                if mob not in fc_with_k:
                    fc_with_k[mob] = ead_series.copy()
                else:
                    fc_with_k[mob] = fc_with_k[mob].add(ead_series, fill_value=0)
            
            for mob, ead_series in seg_fc_no_k.items():
                if mob not in fc_no_k:
                    fc_no_k[mob] = ead_series.copy()
                else:
                    fc_no_k[mob] = fc_no_k[mob].add(ead_series, fill_value=0)
    
    print(f'✅ Forecast With K (Aggregated): {len(fc_with_k)} MOBs')
    print(f'✅ Forecast No K (Aggregated): {len(fc_no_k)} MOBs')
```

```python
# ============================================================
# TÍNH DEL30_PCT CHO CẢ 3 ĐƯỜNG
# ============================================================

def sum_del30_from_series(ead_series, buckets):
    """Sum DEL30+ from pd.Series (not dict)"""
    total = 0
    for b in buckets:
        if b in ead_series.index:
            total += ead_series[b]
    return total

if selected_key and disb_total > 0:
    results = []
    max_actual_mob = max(mob_dict.keys())
    
    for mob in range(START_MOB, MAX_MOB + 1):
        row = {'MOB': mob}
        
        # Actual (chỉ có đến max_actual_mob)
        if mob <= max_actual_mob and mob in mob_dict:
            actual_ead = mob_dict[mob]
            del30_actual = sum_del30_from_series(actual_ead, BUCKETS_30P)
            row['Actual'] = del30_actual / disb_total
        else:
            row['Actual'] = np.nan
        
        # Forecast No K
        if mob in fc_no_k:
            fc_ead_no_k = fc_no_k[mob]
            del30_no_k = sum_del30_from_series(fc_ead_no_k, BUCKETS_30P)
            row['FC_No_K'] = del30_no_k / disb_total
        else:
            row['FC_No_K'] = np.nan
        
        # Forecast With K
        if mob in fc_with_k:
            fc_ead_with_k = fc_with_k[mob]
            del30_with_k = sum_del30_from_series(fc_ead_with_k, BUCKETS_30P)
            row['FC_With_K'] = del30_with_k / disb_total
        else:
            row['FC_With_K'] = np.nan
        
        results.append(row)
    
    df_compare = pd.DataFrame(results)
    print('\n📊 DEL30+ Rate Comparison (Aggregated by Product):')
    print(df_compare.to_string(index=False))
```

```python
# ============================================================
# MAE / MAPE ANALYSIS
# ============================================================

if selected_key and disb_total > 0:
    # Chỉ tính error cho các MOB có actual
    df_eval = df_compare.dropna(subset=['Actual']).copy()
    
    # MAE
    df_eval['Error_No_K'] = abs(df_eval['FC_No_K'] - df_eval['Actual'])
    df_eval['Error_With_K'] = abs(df_eval['FC_With_K'] - df_eval['Actual'])
    
    # MAPE (avoid division by zero)
    df_eval['MAPE_No_K'] = np.where(df_eval['Actual'] > 0, 
                                    df_eval['Error_No_K'] / df_eval['Actual'] * 100, np.nan)
    df_eval['MAPE_With_K'] = np.where(df_eval['Actual'] > 0, 
                                      df_eval['Error_With_K'] / df_eval['Actual'] * 100, np.nan)
    
    print('📊 Error Metrics by MOB:')
    cols_to_show = ['MOB', 'Actual', 'FC_No_K', 'FC_With_K', 'Error_No_K', 'Error_With_K', 'MAPE_No_K', 'MAPE_With_K']
    print(df_eval[cols_to_show].to_string(index=False))
    
    # Summary
    print('\n📊 Summary:')
    print(f'   MAE (No K):    {df_eval["Error_No_K"].mean()*100:.4f}%')
    print(f'   MAE (With K):  {df_eval["Error_With_K"].mean()*100:.4f}%')
    print(f'   MAPE (No K):   {df_eval["MAPE_No_K"].mean():.2f}%')
    print(f'   MAPE (With K): {df_eval["MAPE_With_K"].mean():.2f}%')
    
    # Which is better?
    if df_eval['Error_With_K'].mean() < df_eval['Error_No_K'].mean():
        print('\n✅ Forecast With K có MAE thấp hơn → K calibration cải thiện accuracy')
    else:
        print('\n⚠️ Forecast No K có MAE thấp hơn → K calibration không cải thiện')
```
